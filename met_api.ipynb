```python
import requests
```

```python
# Поиск по API в галерее НЙорка
params={"q": "Van Gogh", "hasImages": "true"}
base_api_url="https://collectionapi.metmuseum.org/public/collection/v1/search" #это енд поинт из справочника апи Музея
search = requests.get(base_api_url, params=params).json() # json сразу интерпертировал выдачу и вернул как справочник (ключ- значение))
```
